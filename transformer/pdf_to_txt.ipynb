The result **pdf_to_text.ipynb** means that py is running within the tranformer/ directory and thus we have to move one directory upward before accessing the documents/raw

```python
# getting current dierectory
current_directory = Path.cwd()

# moving one level up
project_root = current_directory.parent

# navigating to folder of pdfs
raw_folder = project_root / "documents" / "raw"

# finding all the pdf docs
pdf_files = sorted(raw_folder.glob("*.pdf"))
```

```python
# printing the number of pdfs
print("PDFs found:", len(pdf_files))
```

```python
# display names of PDFs
for pdf in pdf_files:
    print(pdf.name)
```

## Test the first PDF

```python
# Select the first PDF from our list
pdf_path = pdf_files[0]

# Open the PDF using PyMuPDF
document = pymupdf.open(pdf_path)

# Display basic information about the PDF
print("Document:", pdf_path.name)
print("Number of pages:", len(document))

# Access the first page (Python indexing begins at 0)
first_page = document[0]

# Extract the text from that page
text = first_page.get_text("text")

# Display the first 1000 characters
print("\nExtracted text:\n")
print(text[:1000])

# Close the PDF to release resources
document.close()
```

Before processing all the documents, we will first process one as a test

```python
# Open the selected PDF
document = pymupdf.open(pdf_path)

# Create an empty string to store the extracted text
full_text = ""

# Iterate through every page of the PDF
for page in document:

    # Extract the text of the current page
    page_text = page.get_text("text")

    # Append the extracted text to our string
    full_text += page_text + "\n"

# Close the PDF
document.close()

# Display extraction statistics
print("Total characters:", len(full_text))
print("Total words:", len(full_text.split()))

# Preview a section beyond the opening metadata
print("\nText preview:\n")
print(full_text[5000:6000])
```
